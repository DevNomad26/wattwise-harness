### 🧪 What This Notebook Evaluates:
1. **Evaluation 1: Tariff & Math Reasoning Benchmark** (`eval/run_eval.py`)
   - Domestic tariff slab calculations across Indian states (Maharashtra, Gujarat, Rajasthan, Delhi, Karnataka).
   - Arithmetic precision (`calculator.py` MCP tool vs. LLM mental math).

2. **Evaluation 2: Bill Photo Vision Benchmark** (`eval/run_vision_eval.py`)
   - State identification and DISCOM recognition directly from bill images.
   - Structured `BillData` extraction (units, amount due, meter readings) vs unassisted unstructured text.
   - Algorithmic integrity & sanity checks (`current - previous == units`).

## 🚀 Step 2: Start Ollama Daemon & Pull Models

```python
import subprocess, time, urllib.request, json, os

# Clean up any previous Ollama processes
os.system("pkill -9 -f ollama")
time.sleep(2)

# Start Ollama background server
subprocess.Popen(["ollama", "serve"])
time.sleep(4)

# Pull the evaluation model onto GPU
!ollama pull qwen3.5:4b

# Verify server is online and model is loaded
with urllib.request.urlopen("http://localhost:11434/api/tags", timeout=5) as res:
    models = [m["name"] for m in json.loads(res.read())["models"]]
    print("\n✅ Ollama is ONLINE with models:", models)
```
